```python
#%%
def getModel(model_name, dataset_conf, from_logits = False):

    n_classes = dataset_conf.get('n_classes')
    n_channels = dataset_conf.get('n_channels')
    in_samples = dataset_conf.get('in_samples')

    # Select the model
    if(model_name == 'RockNet'):
        # Train using the proposed ATCNet model: https://ieeexplore.ieee.org/document/9852687
        model = models.RockNet_(
            # Dataset parameters
            n_classes = n_classes,
            in_chans = n_channels,
            in_samples = in_samples,
            # Sliding window (SW) parameter
            n_windows = 5,
            # Attention (AT) block parameter
            attention = 'mha', # Options: None, 'mha','mhla', 'cbam', 'se'
            # Convolutional (CV) block parameters
            eegn_F1 = 16,
            eegn_D = 2,
            eegn_kernelSize = 64,
            eegn_poolSize = 7,
            eegn_dropout = 0.5,
            # Temporal convolutional (TC) block parameters
            tcn_depth = 2,
            tcn_kernelSize = 4,
            tcn_filters = 64,
            tcn_dropout = 0.3,
            tcn_activation='elu',
            )
    elif model_name == 'RatioWaveNet2':
        model = models.RatioWaveNet2(
            n_classes = n_classes,
            in_chans = n_channels,
            in_samples = in_samples,
            n_windows = 5,
            attention = 'mha',
            eegn_F1 = 16, 
            eegn_D = 2,
            eegn_kernelSize = 64, 
            eegn_poolSize = 7, 
            eegn_dropout = 0.3,
            tcn_depth = 4, 
            tcn_kernelSize = 5, 
            tcn_filters = 48,
            tcn_dropout = 0.3, 
            tcn_activation='elu',
            # --- RDWT ---
            use_rdwt = True,
            rdwt_levels = 4,
            rdwt_wavelet = 'db4',
            rdwt_init_dilations = (1.5, 5/3, 7/4, 9/5),
            rdwt_kernel_len = 16,
            rdwt_soft_threshold = True,
        )
    elif(model_name == 'RockNetA'):
        model = models.RockNetA(
            n_classes = n_classes,
            in_chans = n_channels,
            in_samples = in_samples,
            n_windows = 5,
            attention = 'mha',
            eegn_F1 = 16,
            eegn_D = 2,
            eegn_kernelSize = 64,
            eegn_poolSize = 7,
            eegn_dropout = 0.3,
            tcn_depth = 2,
            tcn_kernelSize = 4,
            tcn_filters = 32,
            tcn_dropout = 0.3,
            tcn_activation='elu',
        )
    elif(model_name == 'RockNetB'):
        model = models.RockNetB(
            n_classes = n_classes,
            in_chans = n_channels,
            in_samples = in_samples,
            n_windows = 5,
            attention = 'mha',
            eegn_F1 = 16,
            eegn_D = 2,
            eegn_kernelSize = 64,
            eegn_poolSize = 7,
            eegn_dropout = 0.5,
            tcn_depth = 2,
            tcn_kernelSize = 4,
            tcn_filters = 32,
            tcn_dropout = 0.3,
            tcn_activation='elu',
        )
    elif(model_name == 'RockNetC'):
        model = models.RockNetC(
            n_classes = n_classes,
            in_chans = n_channels,
            in_samples = in_samples,
            n_windows = 5,
            attention = 'mha',
            eegn_F1 = 16,
            eegn_D = 2,
            eegn_kernelSize = 64,
            eegn_poolSize = 7,
            eegn_dropout = 0.5,
            tcn_depth = 2,
            tcn_kernelSize = 4,
            tcn_filters = 32,
            tcn_dropout = 0.3,
            tcn_activation='elu',
        )
    elif(model_name == 'RockNetX'):
        model = models.RockNetX(
            n_classes = n_classes,
            in_chans = n_channels,
            in_samples = in_samples,
            n_windows = 5,
            attention = 'transformer',
            eegn_F1 = 16,
            eegn_D = 2,
            eegn_kernelSize = 64,
            eegn_poolSize = 7,
            eegn_dropout = 0.5,
            tcn_depth = 3,
            tcn_kernelSize = 4,
            tcn_filters = 64,
            tcn_dropout = 0.3,
            tcn_activation='gelu',
        )
    elif(model_name == 'RockNetY'):
        model = models.RockNetY(
            n_classes = n_classes,
            in_chans = n_channels,
            in_samples = in_samples,
            n_windows = 5,
            attention = 'mha',
            eegn_F1 = 16,
            eegn_D = 2,
            eegn_kernelSize = 64,
            eegn_poolSize = 7,
            eegn_dropout = 0.5,
            tcn_depth = 2,
            tcn_kernelSize = 4,
            tcn_filters = 32,
            tcn_dropout = 0.3,
            tcn_activation='elu',
        )
    elif(model_name == 'RockNetZ'):
        model = models.RockNetZ(
            n_classes = n_classes,
            in_chans = n_channels,
            in_samples = in_samples,
            n_windows = 5,
            attention = 'mha',
            eegn_F1 = 16,
            eegn_D = 2,
            eegn_kernelSize = 64,
            eegn_poolSize = 7,
            eegn_dropout = 0.5,
            tcn_depth = 2,
            tcn_kernelSize = 4,
            tcn_filters = 32,
            tcn_dropout = 0.3,
            tcn_activation='relu',
        )
    elif(model_name == 'ATCNet'):
        # Train using the proposed ATCNet model: https://ieeexplore.ieee.org/document/9852687
        model = models.ATCNet_(
            # Dataset parameters
            n_classes = n_classes,
            in_chans = n_channels,
            in_samples = in_samples,
            # Sliding window (SW) parameter
            n_windows = 5,
            # Attention (AT) block parameter
            attention = 'mha', # Options: None, 'mha','mhla', 'cbam', 'se'
            # Convolutional (CV) block parameters
            eegn_F1 = 16,
            eegn_D = 2,
            eegn_kernelSize = 64,
            eegn_poolSize = 7,
            eegn_dropout = 0.3,
            # Temporal convolutional (TC) block parameters
            tcn_depth = 2,
            tcn_kernelSize = 4,
            tcn_filters = 32,
            tcn_dropout = 0.3,
            tcn_activation='elu',
            )
    elif(model_name == 'TCNet_Fusion'):
        # Train using TCNet_Fusion: https://doi.org/10.1016/j.bspc.2021.102826
        model = models.TCNet_Fusion(n_classes = n_classes, Chans=n_channels, Samples=in_samples)
    elif(model_name == 'EEGTCNet'):
        # Train using EEGTCNet: https://arxiv.org/abs/2006.00622
        model = models.EEGTCNet(n_classes = n_classes, Chans=n_channels, Samples=in_samples)
    elif(model_name == 'EEGNet'):
        # Train using EEGNet: https://arxiv.org/abs/1611.08024
        model = models.EEGNet_classifier(n_classes = n_classes, Chans=n_channels, Samples=in_samples)
    elif(model_name == 'EEGNeX'):
        # Train using EEGNeX: https://arxiv.org/abs/2207.12369
        model = models.EEGNeX_8_32(n_timesteps = in_samples , n_features = n_channels, n_outputs = n_classes)
    elif(model_name == 'DeepConvNet'):
        # Train using DeepConvNet: https://doi.org/10.1002/hbm.23730
        model = models.DeepConvNet(nb_classes = n_classes , Chans = n_channels, Samples = in_samples)
    elif(model_name == 'ShallowConvNet'):
        # Train using ShallowConvNet: https://doi.org/10.1002/hbm.23730
        model = models.ShallowConvNet(nb_classes = n_classes , Chans = n_channels, Samples = in_samples)
    elif(model_name == 'MBEEG_SENet'):
        # Train using MBEEG_SENet: https://www.mdpi.com/2075-4418/12/4/995
        model = models.MBEEG_SENet(nb_classes = n_classes , Chans = n_channels, Samples = in_samples)

    else:
        raise Exception("'{}' model is not supported yet!".format(model_name))

    return model
```

```python
# ==== CELL 1: utility ====
import os, random, numpy as np, tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, cohen_kappa_score

def set_global_seed(seed: int):
    os.environ["PYTHONHASHSEED"] = str(seed)
    random.seed(seed)
    np.random.seed(seed)
    tf.keras.utils.set_random_seed(seed)

def _log_d_from_scales(scales):
    # s = 1 + softplus(log_d)  ->  log_d = softplus^{-1}(s-1) = log(exp((s-1)) - 1)
    s = np.asarray(scales, dtype=np.float32)
    return np.log(np.expm1(s - 1.0) + 1e-6).astype(np.float32)

def reset_rdwt_scales_to_literature(model, init_scales=(1.5, 5/3, 7/4, 9/5), layer_name='rdwt'):
    """
    Forza i fattori di dilatazione dell'RDWT ai valori di letteratura PRIMA di ogni run.
    
    """
    init_scales = np.array(init_scales, dtype=np.float32)
    for layer in model.layers:
        if layer.name == layer_name and hasattr(layer, "log_d"):
            L = int(layer.levels) if hasattr(layer, "levels") else len(init_scales)
            d = init_scales[:L]
            layer.log_d.assign(_log_d_from_scales(d))
            # opzionale: azzera soglie iniziali se presenti
            if hasattr(layer, "th_raw"):
                layer.th_raw.assign(tf.zeros_like(layer.th_raw))
            if hasattr(layer, "tau"):
                layer.tau.assign(tf.zeros_like(layer.tau))
            break  # supponiamo un solo layer 'rdwt'
```

```python
# ==== CELL 2: worst-subject finder (per dataset) ====
def quick_val_acc_for_subject(dataset_conf, train_conf, subject_idx, epochs_quick=50, seed=123,
                              init_scales=(1.5, 5/3, 7/4, 9/5)):
    set_global_seed(seed)

    # Dati
    X_train, _, y_train_onehot, _, _, _ = get_data(
        dataset_conf['data_path'], subject_idx, dataset_conf['name'],
        LOSO=dataset_conf['LOSO'], isStandard=dataset_conf['isStandard']
    )
    X_train, X_val, y_train_onehot, y_val_onehot = train_test_split(
        X_train, y_train_onehot, test_size=0.2, random_state=42
    )
    X_train = apply_preprocessing(X_train, preprocessing=dataset_conf['signal_preprocessing'])
    X_val   = apply_preprocessing(X_val,   preprocessing=dataset_conf['signal_preprocessing'])

    # Modello nuovo per questo subject
    from_logits = train_conf.get('from_logits', False)
    model = getModel(train_conf['model'], dataset_conf, from_logits)
    # Forza RDWT alle dilatazioni di letteratura
    reset_rdwt_scales_to_literature(model, init_scales=init_scales, layer_name='rdwt')

    model.compile(
        loss=tf.keras.losses.CategoricalCrossentropy(from_logits=from_logits),
        optimizer=tf.keras.optimizers.Adam(learning_rate=train_conf['lr']),
        metrics=['accuracy']
    )
    model.fit(X_train, y_train_onehot,
              validation_data=(X_val, y_val_onehot),
              epochs=epochs_quick, batch_size=train_conf['batch_size'], verbose=0)

    y_pred = model.predict(X_val, verbose=0)
    if from_logits: y_pred = tf.nn.softmax(y_pred).numpy()
    y_pred = y_pred.argmax(axis=-1)
    y_true = y_val_onehot.argmax(axis=-1)
    return accuracy_score(y_true, y_pred)

def find_worst_subject(dataset_conf, train_conf, epochs_quick=50, seed=123, init_scales=(1.5, 5/3, 7/4, 9/5)):
    scores = []
    for sid in range(dataset_conf['n_sub']):
        acc = quick_val_acc_for_subject(dataset_conf, train_conf, sid, epochs_quick, seed, init_scales)
        scores.append((sid, float(acc)))
    scores.sort(key=lambda t: t[1])  # crescente → peggiore in testa
    worst_sid = scores[0][0]
    return worst_sid, scores
```

```python
# ==== CELL 3: 5-seed runner sul worst subject ====
def run_5_seeds_on_subject(dataset_conf, train_conf, subject_idx, seeds=(1,2,3,4,5),
                           init_scales=(1.5, 5/3, 7/4, 9/5)):
    # Carica split train/test ufficiale del tuo pipeline
    X_train, _, y_train_onehot, X_test, _, y_test_onehot = get_data(
        dataset_conf['data_path'], subject_idx, dataset_conf['name'],
        LOSO=dataset_conf['LOSO'], isStandard=dataset_conf['isStandard']
    )
    # val interna per early dinamico
    X_train, X_val, y_train_onehot, y_val_onehot = train_test_split(
        X_train, y_train_onehot, test_size=0.2, random_state=42
    )

    # Preprocessing 
    X_train = apply_preprocessing(X_train, preprocessing=dataset_conf['signal_preprocessing'])
    X_val   = apply_preprocessing(X_val,   preprocessing=dataset_conf['signal_preprocessing'])
    X_test  = apply_preprocessing(X_test,  preprocessing=dataset_conf['signal_preprocessing'])

    accs, kaps = [], []
    from_logits = train_conf.get('from_logits', False)

    for s in seeds:
        set_global_seed(s)

        # Modello nuovo per ogni seed
        model = getModel(train_conf['model'], dataset_conf, from_logits)
        # Forza RDWT ai fattori di letteratura PRIMA di allenare
        reset_rdwt_scales_to_literature(model, init_scales=init_scales, layer_name='rdwt')

        model.compile(
            loss=tf.keras.losses.CategoricalCrossentropy(from_logits=from_logits),
            optimizer=tf.keras.optimizers.Adam(learning_rate=train_conf['lr']),
            metrics=['accuracy']
        )

        # Allenamento (usa le epoche del tuo train_conf)
        model.fit(X_train, y_train_onehot,
                  validation_data=(X_val, y_val_onehot),
                  epochs=train_conf['epochs'],
                  batch_size=train_conf['batch_size'],
                  verbose=0)

        # Test
        y_prob = model.predict(X_test, verbose=0)
        if from_logits: y_prob = tf.nn.softmax(y_prob).numpy()
        y_hat = y_prob.argmax(axis=-1)
        y_true = y_test_onehot.argmax(axis=-1)

        accs.append(accuracy_score(y_true, y_hat))
        kaps.append(cohen_kappa_score(y_true, y_hat))

    accs = np.array(accs); kaps = np.array(kaps)
    print(f"Soggetto S{subject_idx+1}, seeds {list(seeds)}")
    print("Accuracy: " + ", ".join(f"{a:.3f}" for a in accs) + f"  → mean={accs.mean():.3f}, std={accs.std(ddof=1):.3f}")
    print("Kappa:    " + ", ".join(f"{k:.3f}" for k in kaps) + f"  → mean={kaps.mean():.3f}, std={kaps.std(ddof=1):.3f}")
    return {"accs": accs, "kappas": kaps}
```

```python
# ==== CELL 4: builder dataset_conf/train_conf per ciascun dataset ====
def build_dataset_conf(dataset, signal_preprocessing='none'):
    if dataset == 'BCI2a':
        in_samples = 1125; n_channels = 22; n_sub = 9;  n_classes = 4
        classes_labels = ['Left hand', 'Right hand','Foot','Tongue']
        data_path = 'dataset/'
    elif dataset == 'HGD':
        in_samples = 1125; n_channels = 44; n_sub = 14; n_classes = 4
        classes_labels = ['Right Hand', 'Left Hand','Rest','Feet']
        data_path = 'dataset/'
    elif dataset == 'BCI2b':
        in_samples = 750; n_channels = 3; n_sub = 9; n_classes = 2
        classes_labels = ['Left hand', 'Right hand']
        data_path = 'dataset/'
    else:
        raise ValueError(dataset)

    dataset_conf = {
        'name': dataset, 'n_classes': n_classes, 'cl_labels': classes_labels,
        'n_sub': n_sub, 'n_channels': n_channels, 'in_samples': in_samples,
        'data_path': data_path, 'isStandard': True, 'LOSO': False,
        'signal_preprocessing': signal_preprocessing
    }
    return dataset_conf

def build_train_conf(model_name='RatioWaveNet2', lr=1e-3, batch_size=64, epochs=1500, from_logits=False):
    return {
        'batch_size': batch_size, 'epochs': epochs, 'patience': 150, 'lr': lr, 'n_train': 1,
        'LearnCurves': False, 'from_logits': from_logits, 'model': model_name
    }
```

```python
# ==== CELL 5: orchestration per i tre dataset ====
def do_prof_protocol(signal_preprocessing='none', seeds=(1,2,3,4,5),
                     quick_epochs=50, init_scales=(1.5, 5/3, 7/4, 9/5),
                     model_name='RatioWaveNet2', lr=1e-3, batch_size=64, epochs=1500, from_logits=False,
                     datasets=('BCI2a','HGD','BCI2b')):
    results = {}
    for ds in datasets:
        print("\n" + "="*70)
        print(f"DATASET: {ds}")
        dataset_conf = build_dataset_conf(ds, signal_preprocessing=signal_preprocessing)
        train_conf   = build_train_conf(model_name=model_name, lr=lr, batch_size=batch_size, epochs=epochs, from_logits=from_logits)

        worst_sid, table = find_worst_subject(dataset_conf, train_conf, epochs_quick=quick_epochs, seed=123, init_scales=init_scales)
        print(f"Worst subject (0-based): {worst_sid}  ⇒ S{worst_sid+1}")
        for sid, acc in table:
            print(f"S{sid+1}: {acc:.3f}")

        print(f"[{ds}] 5 run con semi {list(seeds)} su S{worst_sid+1}...")
        out = run_5_seeds_on_subject(dataset_conf, train_conf, subject_idx=worst_sid, seeds=seeds, init_scales=init_scales)

        results[ds] = {"worst_subject": int(worst_sid+1),
                       "accs": out["accs"], "kappas": out["kappas"],
                       "acc_mean": float(out["accs"].mean()), "acc_std": float(out["accs"].std(ddof=1)),
                       "kappa_mean": float(out["kappas"].mean()), "kappa_std": float(out["kappas"].std(ddof=1))}
    return results
```
